# Atributos sin informacion (near zero variance)

```python
def check_nzv(dataframe, threshold=0.95):
    """
    Identifica columnas donde el valor más dominante excede el umbral dado.
    """
    nzv_cols = []
    for col in dataframe.columns:
        # Ignoramos nulos temporalmente para el cálculo de frecuencia de valores presentes
        if dataframe[col].dropna().empty:
            continue
            
        # value_counts con normalize=True devuelve la proporción (0 a 1)
        max_freq = dataframe[col].value_counts(normalize=True).iloc[0]
        
        if max_freq >= threshold:
            nzv_cols.append({'Columna': col, 'Max_Freq': max_freq, 'Top_Value': dataframe[col].mode()[0]})
            
    return pd.DataFrame(nzv_cols).sort_values(by='Max_Freq', ascending=False)

# Aplicamos la función a nuestro dataset
nzv_report = check_nzv(df, threshold=0.98)
print("Variables con Varianza Cercana a Cero (dominancia > 98%):")
display(nzv_report.head(10))
```

```python
# Eliminación de variables NZV
df_clean = df.drop(columns=nzv_report['Columna'].tolist())
```

## 1. Sótanos: ceros estructurales

Criterio: una casa **no tiene sótano** si `TotalBsmtSF == 0`. En esas filas las categóricas pasan a `'Sin_Sotano'` y las numéricas a `0`.

```python
bsmt_cat = ['BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2']
bsmt_num = ['BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', 'BsmtFullBath', 'BsmtHalfBath']

sin_sotano = df_orig['TotalBsmtSF'] == 0
print(f"Casas sin sótano: {sin_sotano.sum()}")

df_b = df_orig.copy()
df_b.loc[sin_sotano, bsmt_cat] = 'Sin_Sotano'
df_b.loc[sin_sotano, bsmt_num] = 0

# Nulos que quedan: casas CON sótano a las que les falta un dato (no son estructurales)
print(df_b.loc[df_b[bsmt_cat].isna().any(axis=1), bsmt_cat + bsmt_num])
```

Quedan 2 nulos que **no** son estructurales (la casa tiene sótano, falta el dato):
- `BsmtExposure` faltante → moda (`'No'`).
- `BsmtFinType2` faltante con `BsmtFinSF2 > 0` → no puede ser `'Unf'`, se usa la moda entre sótanos con segunda área terminada.

```python
df_b['BsmtExposure'] = df_b['BsmtExposure'].fillna(df_b.loc[~sin_sotano, 'BsmtExposure'].mode()[0])

moda_fin2 = df_b.loc[df_b['BsmtFinSF2'] > 0, 'BsmtFinType2'].mode()[0]
df_b['BsmtFinType2'] = df_b['BsmtFinType2'].fillna(moda_fin2)

# Verificación de consistencia
print("Nulos restantes:", df_b[bsmt_cat + bsmt_num].isna().sum().sum())
print("Sin_Sotano con área > 0:", (df_b.loc[df_b['BsmtQual'] == 'Sin_Sotano', bsmt_num] != 0).any(axis=1).sum())
```

## 2. Auditoría de inconsistencias

Para cada par (categórica de la estructura, numérica de la estructura) se buscan dos anomalías:
- **Sin estructura pero con medida**: categórica nula (o `'None'`) y numérica `> 0`.
- **Con estructura pero sin medida**: categórica informada y numérica `== 0`.

```python
pares = {
    'PoolQC': 'PoolArea',
    'FireplaceQu': 'Fireplaces',
    'GarageType': 'GarageArea',
    'MiscFeature': 'MiscVal',
    'MasVnrType': 'MasVnrArea',
}

for cat, num in pares.items():
    sin_estructura = df_orig[cat].isna() | (df_orig[cat] == 'None')
    a = sin_estructura & (df_orig[num] > 0)
    b = ~sin_estructura & (df_orig[num] == 0)
    print(f"{cat:12s} / {num:11s} -> sin estructura con medida: {a.sum():2d} | con estructura sin medida: {b.sum():2d}")
```

```python
mas_none = df_orig['MasVnrType'].isna() | (df_orig['MasVnrType'] == 'None')
anomalias_mas = (mas_none & (df_orig['MasVnrArea'] > 0)) | (~mas_none & (df_orig['MasVnrArea'] == 0))
df_orig.loc[anomalias_mas, ['MasVnrType', 'MasVnrArea']]
```

Las inconsistencias reales están en el revestimiento de mampostería (`MasVnr`):
- **Tipo `'None'` con área de 1 ft²**: ruido de carga → área = `0`.
- **Tipo `'None'` con área grande (288–344 ft²)**: el área es creíble, falta el tipo → moda de los tipos reales (`'BrkFace'`).
- **Tipo informado con área 0**: sin área no hay revestimiento → tipo = `'None'`.
- Los 8 casos con ambos nulos → `'None'` / `0` (nulo estructural).

`MiscFeature` con `MiscVal == 0` (2 casos) se deja como está: la estructura puede existir y valer $0.

```python
df_m = df_orig.copy()
mas_none = df_m['MasVnrType'].isna() | (df_m['MasVnrType'] == 'None')

ruido = mas_none & (df_m['MasVnrArea'] > 0) & (df_m['MasVnrArea'] <= 1)
df_m.loc[ruido, 'MasVnrArea'] = 0

falta_tipo = mas_none & (df_m['MasVnrArea'] > 1)
df_m.loc[falta_tipo, 'MasVnrType'] = df_m.loc[~mas_none, 'MasVnrType'].mode()[0]

df_m.loc[~mas_none & (df_m['MasVnrArea'] == 0), 'MasVnrType'] = 'None'

ambos_nulos = df_m['MasVnrType'].isna() & df_m['MasVnrArea'].isna()
df_m.loc[ambos_nulos, ['MasVnrType', 'MasVnrArea']] = ['None', 0]

# Verificación
mas_none = df_m['MasVnrType'] == 'None'
print("Anomalías restantes:", ((mas_none & (df_m['MasVnrArea'] > 0)) | (~mas_none & (df_m['MasVnrArea'] == 0))).sum())
```

## 3. NZV en numéricas: coeficiente de variación

Se calcula $CV = \sigma / \mu$ excluyendo los ceros (estructurales) y se marca NZV si $CV < 0.05$.

```python
def check_nzv_cv(dataframe, threshold=0.05):
    num = dataframe.select_dtypes('number').drop(columns='Id', errors='ignore')
    res = []
    for col in num.columns:
        s = num[col].replace(0, np.nan).dropna()  # ignoramos ceros estructurales
        if s.empty or s.mean() == 0:
            continue
        res.append({'Columna': col, 'CV': s.std() / s.mean(), 'n_no_cero': len(s)})
    res = pd.DataFrame(res).sort_values('CV')
    return res[res['CV'] < threshold]

check_nzv_cv(df_orig)
```

Las únicas variables con $CV < 0.05$ son **años** (`YrSold`, `YearRemodAdd`, `GarageYrBlt`, `YearBuilt`). Es un artefacto: el CV solo tiene sentido en escalas de razón (con cero absoluto); un año está en escala de intervalo con origen arbitrario, por eso $\mu \approx 2000$ infla el denominador.

No se eliminan. Para evaluarlas correctamente conviene transformarlas en antigüedades (ej. `YrSold - YearBuilt`) y recalcular.

### Transformación a antigüedad

Se pasan los años a antigüedad al momento de la venta. `YrSold` no se transforma porque es la referencia; con solo 5 valores (2006–2010) conviene tratarla como categórica.

Acá el 0 es un valor real (casa vendida el mismo año en que se construyó), así que el CV se calcula **sin** excluir ceros.

```python
antig = pd.DataFrame({
    'Edad': df_orig['YrSold'] - df_orig['YearBuilt'],
    'AniosDesdeRemod': (df_orig['YrSold'] - df_orig['YearRemodAdd']).clip(lower=0),  # 1 caso remodelado post-venta
    'EdadGarage': df_orig['YrSold'] - df_orig['GarageYrBlt'],  # NaN si no hay garage
})

pd.DataFrame({
    'CV_original': [df_orig[c].std() / df_orig[c].mean() for c in ['YearBuilt', 'YearRemodAdd', 'GarageYrBlt']],
    'CV_antiguedad': antig.std() / antig.mean(),
}, index=antig.columns)
```

### Comparación con una versión robusta (RobustScaler)

`RobustScaler` centra con la **mediana** (`center_`) y escala con el **IQR** (`scale_`), así que da los insumos para un CV robusto:

$$CV_{rob} = \frac{IQR / 1.349}{\text{mediana}}$$

El 1.349 hace que, en una normal, $IQR/1.349 = \sigma$, y así el umbral de 0.05 es comparable con el CV clásico.

Ojo: cuando el IQR es 0, `RobustScaler` guarda `scale_ = 1` para no dividir por cero. Hay que detectarlo y corregirlo.

```python
from sklearn.preprocessing import RobustScaler

num = df_orig.select_dtypes('number').drop(columns='Id').replace(0, np.nan)  # sin ceros estructurales; los NaN se ignoran en fit
rs = RobustScaler().fit(num)

iqr = pd.Series(rs.scale_, index=num.columns)
iqr[num.quantile(0.75) == num.quantile(0.25)] = 0  # corrige el scale_=1 que pone sklearn cuando IQR=0

comparacion = pd.DataFrame({
    'CV': num.std() / num.mean(),
    'CV_robusto': (iqr / 1.349) / rs.center_,
})
comparacion['NZV_CV'] = comparacion['CV'] < 0.05
comparacion['NZV_robusto'] = comparacion['CV_robusto'] < 0.05

comparacion[comparacion['NZV_CV'] | comparacion['NZV_robusto']].sort_values('CV_robusto')
```

- **Años**: los dos métodos los marcan, con el mismo artefacto de escala explicado arriba.
- **Conteos discretos** (`KitchenAbvGr`, `HalfBath`, `Fireplaces`, `BsmtFullBath`, `BsmtHalfBath`): solo los marca el robusto. Entre las casas que tienen la estructura, más del 50% tiene el mismo valor (1), así que el IQR es 0 aunque haya casos con 2 o 3. La versión robusta es demasiado brusca para variables discretas con pocos valores; para esas conviene el criterio de frecuencias (`check_nzv`).
- El CV robusto es útil en continuas asimétricas con outliers (ej. `MiscVal`: CV 1.93 contra 0.67 robusto), donde la media y el σ quedan inflados por pocos valores extremos.
